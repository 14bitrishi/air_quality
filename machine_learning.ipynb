```
x = [1, 2, 3]
y = [3, 5, 7]
```

```
w = 2
b = 1
```

```
m = len(x)
```

```
for i in range(m):
    prediction = w * x[i] + b
    print("x =", x[i])
    print("prediction =", prediction)
    print("actual y =", y[i])
    print("error = " , prediction- y[i])
    print("------")
```

```
x = [1, 2, 3]
y = [3, 5, 7]
```

```
w=1
b=2
```

```
m=len(x)
```

```
total_cost=0
```

```
for i in range(m):
  prediction=w*x[i]+b
  error=prediction - y[i]
  cost=error**2
  total_cost+=cost
  print("prediction = ",prediction)
  print("error=" , error)
  print("cost= " , cost)
  print("--------------------------------")
final_cost=total_cost / (2*m)
print("final cost =",final_cost)
```
