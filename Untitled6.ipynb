```python
x = [1,2,3]
y = [3,5,7]
```

```python
w = 2
b = 1
```

```python
m=len(x)
```

```python
for i in range(m):
  prediction=w*x[i]+b
  print("x=" , x[i])
  print("prediction =" , prediction)
  print("actual y =" , y[i])
  print("error = " , prediction- y[i])
  print("------------------")
```
